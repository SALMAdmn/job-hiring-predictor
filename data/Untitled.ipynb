```python
import pandas as pd
import numpy as np

# Reproductibilité
np.random.seed(42)

n_rows = 5000

# ─────────────────────────────────────────────
# 1. VARIABLES MAROCAINES
# ─────────────────────────────────────────────
filieres = ['GI', 'GINF', 'Data Science', 'Réseaux & Télécoms', 'Développement Web', 'Cybersécurité']
ecoles = ['ENSA', 'EST', 'FST', 'Faculté (FS)', 'OFPPT', 'Privée (EMSI/IGA)']
villes = ['Casablanca', 'Rabat', 'Tanger', 'Marrakech', 'Agadir', 'Fès', 'Kenitra']

# ─────────────────────────────────────────────
# 2. GÉNÉRATION DES DONNÉES
# ─────────────────────────────────────────────
data = {
    'filiere': np.random.choice(filieres, n_rows),
    'type_ecole': np.random.choice(ecoles, n_rows, p=[0.25, 0.15, 0.15, 0.15, 0.20, 0.10]),
    'ville': np.random.choice(villes, n_rows),
    'moyenne': np.round(np.random.uniform(10, 18, n_rows), 2),
    'stages': np.random.randint(0, 5, n_rows),
    'niveau_francais': np.random.randint(4, 11, n_rows),
    'niveau_anglais': np.random.randint(3, 11, n_rows),
    'niveau_programmation': np.random.randint(3, 11, n_rows),
    'niveau_algorithme': np.random.randint(3, 11, n_rows),
    'projets': np.random.randint(0, 8, n_rows),
    'certifications': np.random.randint(0, 4, n_rows),
    'participation_hackathon': np.random.choice([0, 1], n_rows, p=[0.7, 0.3]),
    'open_source_contributions': np.random.choice([0, 1], n_rows, p=[0.85, 0.15]),
    'soft_skills': np.random.randint(4, 11, n_rows),
    'linkedin_profile': np.random.choice([0, 1], n_rows, p=[0.3, 0.7])
}

df = pd.DataFrame(data)

# ─────────────────────────────────────────────
# 3. SCORE DE RECRUTEMENT (AMÉLIORÉ)
# ─────────────────────────────────────────────
score = (
    df['moyenne'] * 2 +
    df['stages'] * 15 +
    df['niveau_francais'] * 8 +
    df['niveau_anglais'] * 10 +
    df['niveau_programmation'] * 12 +
    df['niveau_algorithme'] * 10 +
    df['projets'] * 8 +
    df['certifications'] * 5 +
    df['participation_hackathon'] * 10 +
    df['open_source_contributions'] * 12 +
    df['soft_skills'] * 7 +
    df['linkedin_profile'] * 5
)

# Normalisation
prob_recrutement = (score - score.min()) / (score.max() - score.min())

# 🔥 Ajout de bruit (IMPORTANT pour réalisme)
noise = np.random.normal(0, 0.1, n_rows)

df['recrute'] = ((prob_recrutement + noise) > 0.55).astype(int)

# ─────────────────────────────────────────────
# 4. SALAIRE ESTIMÉ (MAROC)
# ─────────────────────────────────────────────
def estimer_salaire(row):
    if row['recrute'] == 0:
        return 0

    base_salaires = {
        'ENSA': 8000,
        'EST': 5500,
        'FST': 7000,
        'Faculté (FS)': 6000,
        'OFPPT': 5000,
        'Privée (EMSI/IGA)': 8500
    }

    salaire = base_salaires[row['type_ecole']]

    # Bonus compétences
    salaire += row['niveau_anglais'] * 200
    salaire += row['stages'] * 500
    salaire += row['certifications'] * 400
    salaire += row['open_source_contributions'] * 500

    # Ville (coût de vie)
    if row['ville'] in ['Casablanca', 'Rabat']:
        salaire += 1000

    return int(salaire)

df['salaire_estime_dh'] = df.apply(estimer_salaire, axis=1)

# ─────────────────────────────────────────────
# 5. SAUVEGARDE
# ─────────────────────────────────────────────
df.to_csv('maroc_tech_students_dataset.csv', index=False)

print("✅ Dataset généré avec succès !")
print(df.head())
```
